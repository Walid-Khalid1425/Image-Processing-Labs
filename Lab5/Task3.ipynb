```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Read the image
image = cv2.imread('Parrot.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Step 2: Convert to float32
image_float = image.astype(np.float32) / 255.0

# Step 3: Apply Laplacian filter
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# Step 4: Sharpen the image
sharpened = np.clip(image_float - laplacian, 0, 1)

# Step 5: Plot
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].imshow(image)
ax[0].set_title("Original Image")
ax[0].axis("off")
ax[1].imshow(np.clip(laplacian, 0, 1))
ax[1].set_title("Laplacian Filter")
ax[1].axis("off")
ax[2].imshow(sharpened)
ax[2].set_title("Sharpened Image")
ax[2].axis("off")
plt.tight_layout()
plt.show()
```
