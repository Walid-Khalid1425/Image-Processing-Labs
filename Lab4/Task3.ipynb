```python
import matplotlib.pyplot as plt
from skimage import data, exposure
```

```python
source = data.rocket()     # image to modify
reference = data.chelsea() # target histogram
```

```python
matched = exposure.match_histograms(source, reference, channel_axis=-1)
```

```python
fig, axes = plt.subplots(1,3, figsize=(15,5))

axes[0].imshow(source)
axes[0].set_title('Source (Rocket)')
axes[0].axis('off')

axes[1].imshow(reference)
axes[1].set_title('Reference (Chelsea)')
axes[1].axis('off')

axes[2].imshow(matched)
axes[2].set_title('Matched Image')
axes[2].axis('off')

plt.show()
```
