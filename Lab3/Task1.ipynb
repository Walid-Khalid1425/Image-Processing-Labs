```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("Pic.png")
plt.imshow(img)
plt.axis("off")
```

```python
#increase the image size
bigger = cv2.resize(img, None, fx=2, fy=2)
plt.figure(figsize=(6,3))
plt.subplot(1,2,1)
plt.imshow(img[:,:,::-1])
plt.title("Original")
plt.subplot(1,2,2)
plt.imshow(bigger[:,:,::-1])
plt.title("Bigger")

plt.show()
```

```python
#rotate 120 degrees
rows, cols = img.shape[:2]
M = cv2.getRotationMatrix2D((cols/2, rows/2), 120, 1)
rotated = cv2.warpAffine(img, M, (cols, rows))

plt.figure(figsize=(6,3))
plt.subplot(1,2,1)
plt.imshow(img[:,:,::-1])
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(rotated[:,:,::-1])
plt.title("Rotated")
plt.axis("off")

plt.show()
```

```python
#Shearing
shear_M = np.float32([[1, 0.3, 0],
                      [0, 1,   0]])
sheared = cv2.warpAffine(img, shear_M, (cols, rows))

plt.figure(figsize=(6,3))
plt.subplot(1,2,1)
plt.imshow(img[:,:,::-1])
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(sheared[:,:,::-1])
plt.title("Sheared")
plt.axis("off")

plt.show()
```
